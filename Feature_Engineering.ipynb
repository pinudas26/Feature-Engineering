2) What is correlation? What does negative correlation mean?
   - In Python, correlation refers to the statistical analysis of the relationship or dependency between two variables.
   - It measures the strength and direction of the relationship, helping to understand how changes in one variable are associated with changes in another.
   - Python libraries like SciPy, NumPy, and pandas provide tools to calculate various correlation coefficients, including the Pearson correlation coefficient, which measures linear relationships.
   - A negative correlation means that two variables tend to move in opposite directions. When one variable increases, the other variable tends to decrease, and vice versa.
   - This relationship is also known as an inverse correlation.

3) Define Machine Learning. What are the main components in Machine Learning?
   - Machine learning is a field within AI that focuses on creating algorithms and models that allow computers to learn from data and make predictions or decisions without being explicitly programmed for each task.
  - It involves training computer systems to learn from data, identify patterns, and use those patterns to make predictions on new, unseen data.
  - ML can be used for a wide range of applications, including image recognition, natural language processing, fraud detection, and more.
  - Here are the key machine learning lifecycle components:
    - Representation
      - This refers to the way knowledge is represented for ML purposes. Some examples include decision trees, sets of rules, instances, graphical models, neural networks, support vector machines, model ensembles, and various others.
    - Abstraction
      - Abstraction simplifies the representation of a problem, allowing for more efficient problem-solving with reduced memory and computation requirements.
      - Examples of data abstraction are decreasing the spatial and temporal resolution or dividing continuous variables into meaningful ranges that align with specific goals.
    - Evaluation
      - Every ML project needs a method for evaluating hypotheses. Some examples are accuracy, prediction and recall, squared error, KL divergence (relative entropy), and others.
    - Generalization
      - Generalization is crucial for a model to effectively handle new, unfamiliar data that comes from the same distribution as the data used to train the model.
      - It allows teams to gain a deeper understanding of overfitting and assess the quality of a model.
    - Data Storage
      - This one might easily get forgotten among the components of machine learning, but where your data resides is very important.
      - Common storage solutions for machine learning include object storage, distributed file systems, and cloud-based storage.

5) What are continuous and categorical variables?
   - Continuous Variables:
     - Definition:
       - Continuous variables can assume an infinite number of values within a given interval.
     - Examples:
       - Height, weight, temperature, time, distance, and blood pressure.
     - Characteristics:
       - They are usually measured and can have values with decimal places or fractions.
     - Analysis:
       - Statistical techniques like regression and t-tests are often used with continuous variables.
   - Categorical variables:
     - Definition: Categorical variables represent distinct categories or groups.
     - Examples: Gender (male/female), race, marital status, blood type, and color.
     - Characteristics: They are descriptive and do not have numerical values, but may be represented by numbers (e.g., 1 for male, 2 for female).
     - Analysis: Techniques like chi-square tests and analysis of variance are used with categorical variables.

15) What is causation? Explain difference between correlation and causation with an example.
  - Causation means one event directly causes another, while correlation indicates a relationship between two events without necessarily implying cause and effect.
   - For example, eating ice cream and getting sunburned are correlated (both happening more often in summer), but ice cream consumption doesn't cause sunburns; the relationship is due to a third factor: sunny weather.

16) What is an Optimizer? What are different types of optimizers? Explain each with an example.
  - In machine learning, an optimizer is an algorithm that adjusts a model's parameters (like weights and biases) to minimize a loss function, guiding the model towards better predictions.
  - It essentially refines the model during training, iteratively updating its parameters to achieve a desired level of accuracy.
  - Common types include Gradient Descent, Stochastic Gradient Descent (SGD), and variants like SGD with Momentum, Mini-batch Gradient Descent, RMSprop, Adam, and others.
  - These algorithms differ in how they update model parameters based on the loss function's gradient.
  - Example of Gradient Descent:
    - In linear regression, it iteratively updates weights to minimize the mean squared error between predicted and actual values.
  - Example of Stochastic Gradient Descent:
    - In a large dataset like image classification, SGD updates weights for each image individually, leading to faster iterations.
  - Example of Mini-batch Gradient Descent:
    - In training a neural network, it updates weights using batches of 32, 64, or 128 images, finding a balance between speed and accuracy.
  - Example of SGD with Momentum:
    - In training a deep learning model, momentum helps the algorithm move through valleys in the loss landscape more efficiently, preventing it from getting stuck.
  - Example of  RMSprop:
    - In training a recurrent neural network for sequence prediction, RMSprop can help prevent the learning rate from becoming too small or too large for certain parameters, ensuring stability.
  - Example of Adam (Adaptive Moment Estimation):
    - In training a large language model, Adam often provides a good balance between convergence speed and stability, making it a popular choice.
  - Example of Adagrad:
    - In sparse datasets like text classification, Adagrad can help the model focus on important words and features, even if they appear infrequently.
  - Example of AdaDelta:
    - In online learning scenarios where the dataset is continuously updated, AdaDelta can adapt to the changing data distribution more effectively.
